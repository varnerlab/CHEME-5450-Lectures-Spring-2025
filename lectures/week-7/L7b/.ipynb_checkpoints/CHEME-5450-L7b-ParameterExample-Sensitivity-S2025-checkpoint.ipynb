# L7b: Local and Global Sensitivity Analysis

---

In this lecture, we will introduce local and global sensitivity analysis. Sensitivity analysis will help us understand how the model output changes with respect to changes in the model parameters. It will also give us some insight into the identifiability of the model parameters. The key ideas of this lecture are:
* __Local sensitivity analysis__ is a method to understand how the model output changes with respect to changes in the model parameters _locally_, i.e., in a small region of some $\theta_{\circ}$, one at a time. It involves calculating the partial derivative of the model output with respect to each parameter. These values give us insight into the identifiability of the model parameters and a nice linkage to the design of experiments.
* __Global sensitivity analysis__ is a method for understanding how the model output changes with respect to large changes in the model parameters. It involves sampling the model's response to parameter changes over a large range of values, where parameters can change jointly. This method helps understand the correlation between the model parameters and identify the most influential parameters on the model output.
* __Parameter identifiability__ refers to the ability to uniquely determine the values of a model's parameters from available data, encompassing both _structural identifiability_, a theoretical property ensuring distinct parameter sets yield different model outputs, and _practical identifiability_ which considers the quality and quantity of the data used for estimation.

## Local Sensitivity Analysis
Local sensitivity analysis is a method used to evaluate how small changes in input parameters affect the output of a model, typically around a specific reference point or base case scenario. This approach helps understand the impact of individual parameters but has limitations, particularly in nonlinear models where interactions between parameters can be underestimated. Suppose we have an ordinary differential equation model, e.g., a system of mRNA and protein balances, of the form:
$$
\begin{align*}
\dot{x} & = f(x, u, \theta)\\
y & = g(x, \theta)\\
x(0) & = x_0
\end{align*}
$$
where $x \in \mathbb{R}^n$ is the state vector (e.g., mRNA and protein concentrations), $u \in \mathbb{R}^m$ is the input vector, $\theta \in \mathbb{R}^p$ is the (unknown) parameter vector, $y \in \mathbb{R}^q$ is the output vector, and $f$ and $g$ are the state and output functions, respectively.

#### Definition: Local Output Sensitivity
Let the ouput $y\in\mathbb{R}^{q}$ vector from the general model above be given by:
$$
\begin{align*}
y_1 & = g_1(x, \theta)\\
y_2 & = g_2(x, \theta)\\
& \vdots\\
y_q & = g_q(x, \theta)
\end{align*}
$$
where $g_i(x, \theta)$ is the $i$ th output function. Then, the local output sensitivity of the model output $y$ with respect to the parameter $\theta_i$ (at time $t$) is defined as:
$$
\frac{\partial y_{i}}{\partial \theta_j} = \frac{\partial g_{i}(x, \theta)}{\partial \theta_j} + \sum_{k=1}^{n} \frac{\partial g_{i}(x, \theta)}{\partial x_k}\left(\frac{\partial x_k}{\partial \theta_j}\right)\quad \text{for } i = 1, 2, \ldots, q\,\text{ and } j = 1, 2, \ldots, p
$$
where the partial derivatives are evaluated at the reference point $(x_{\circ}, \theta_{\circ})$. The first term on the right-hand side of the equation is the direct effect of the parameter $\theta_j$ on the output $y_i$, while the second term is the indirect effect of the parameter $\theta_j$ on the output $y_i$ through the state variables $x$. 

#### Definition: Local State Sensitivity
In the case where we can measure the states directly, i.e., $y = x$, the local output sensitivity of the model output $y$ with respect to the parameter $\theta_i$ is equal to the _state sensitivity_ of the model output $y$ with respect to the parameter $\theta_i$ (at time $t$):
$$
\frac{\partial y_{i}}{\partial \theta_j} = \frac{\partial x_{i}}{\partial \theta_j}\quad \text{for } i = 1, 2, \ldots, n\,\text{ and } j = 1, 2, \ldots, p
$$
where the partial derivatives are evaluated at the reference point $(x_{\circ}, \theta_{\circ})$.

## Sensitivity and Parameter Identifiability
The local sensitivity analysis can be used to understand the identifiability of the model parameters. A parameter is said to be identifiable if it can be uniquely estimated from the model output. Generally, a parameter is identifiable if the model output is sensitive to changes in it. If the model output is not sensitive to changes in the parameter, then the parameter is considered unidentifiable.

There is a simple method that can be used to determine the identifiability of the model parameters using the local sensitivity analysis developed by:
* [Yao, K.Z., Shaw, B.M., Kou, B., McAuley, K.B., and Bacon, D.W.: Modeling ethylene/butene copolymerization with multi-site catalysts: parameter estimability and experimental design, Polym. React. Eng., 2003, 11, pp. 563 – 588](https://www.tandfonline.com/doi/full/10.1081/PRE-120024426)

#### Algorithm: Identifiability Analysis
This algorithm was [reproduced from the Appendix of the Gadkat et al. 2005](https://digital-library.theiet.org/doi/abs/10.1049/sb%3A20045029). The algorithm is as follows:

1. Calculate the local sensitivity of the model output with respect to each parameter for time $t = 1,2,3,\ldots, T$. Each time point will be an $n \times p$ matrix of sensitivities. Stack these matrices to form a $n\cdot{T}\times p$ matrix of sensitivities denoted as $\bar{\mathbf{Z}}$.
2. Calculate the magnitude of each column (sum of squares of the elements) of the matrix $\bar{\mathbf{Z}}$. The parameter corresponding to the column with the maximum magnitude is the first identifiable parameter. This column is called $X_{L}$.
3. Calculate the prediction of the full sensitivity matrix $\bar{\mathbf{Z}}$ using the first identifiable parameter $X_{L}$, i.e., $\hat{\mathbf{Z}} = X_{L}(X_{L}^{\top}X_{L})^{-1}X_{L}^{\top}\bar{\mathbf{Z}}$.
4. Calculate the residual matrix $\mathbf{R} = \bar{\mathbf{Z}} - \hat{\mathbf{Z}}$.
5. Calculate the magnitude of each column (sum of squares of the elements) of the residual matrix $\mathbf{R}$. The parameter corresponding to the column with the maximum magnitude is the _next identifiable parameter_. 
6. Augment the matrix $X_{L}$ with the column corresponding to the next identifiable parameter. Call this matrix $X_{L+1}$.
7. Repeat steps 3-6 until the _maximum magnitude_ of the residual matrix $\mathbf{R}$ is less than a specified threshold.

For more details on parameter identifiability (including additional methods), see:
* [van Noort M, Ruppert M, DeJongh J, Marostica E, Bosch R, Mešić E, Snelder N. Navigating the landscape of parameter identifiability methods: A workflow recommendation for model development. CPT Pharmacometrics Syst Pharmacol. 2024 Jul;13(7):1170-1179. doi: 10.1002/psp4.13148. Epub 2024 May 7. PMID: 38715385; PMCID: PMC11247124.](https://pmc.ncbi.nlm.nih.gov/articles/PMC11247124/)

### How are identifiability and sensitivity analysis related to the design of experiments?
The local sensitivity analysis can design experiments that are sensitive to the model parameters. The idea is to design experiments that maximize the sensitivity of the model output to the model parameters. 

The [Cramer-Rao theorem](https://en.wikipedia.org/wiki/Cramér–Rao_bound) and the [Fisher Information Matrix (FIM)](https://en.wikipedia.org/wiki/Fisher_information) are used to determine a measurement set such that the estimated parameters have minimum variance. The FIM is given by:
$$
\text{FIM} = \mathbf{Z}^{\top}\mathbf{W}\mathbf{Z}
$$
where $\mathbf{Z}$ is the sensitivity matrix of the model output with respect to the _indentifiable_ model parameters, and $\mathbf{W}$ is the weight matrix. The weight matrix is typically the inverse of the measurement error covariance matrix. The FIM can be used to determine the optimal measurement set that maximizes the sensitivity of the model output to the model parameters. In cases when the measurement error is not correlated, the FIM can be used to determine the lower bound for the variance of identifiable parameters:
$$
\text{Var}(\hat{\theta}_{i}) \geq (\text{FIM}^{-1})_{ii}
$$
where $\hat{\theta}_{i}$ is the estimated value of the $i$ th _indentifiable_ parameter. 

Let's look at an example to illustrate the identifiability and its usefulness for experimental design.
* [Hu CY, Varner JD, Lucks JB. Generating Effective Models and Parameters for RNA Genetic Circuits. ACS Synth Biol. 2015 Aug 21;4(8):914-26. doi: 10.1021/acssynbio.5b00077. Epub 2015 Jul 2. PMID: 26046393.](https://pubmed.ncbi.nlm.nih.gov/26046393/)

## Global Sensitivity Analysis
Global sensitivity analysis quantifies how the uncertainty in a model's output can be attributed to different sources of uncertainty in the input variables, considering their entire range of variation and potential interactions. This approach helps identify the most influential variables and understand their contributions to the overall uncertainty in the model's output, enabling model simplification and optimization.

Here are some examples of global sensitivity analysis methods:
* __Sobol Method__: This is a _variance-based approach_ that quantifies each input variable's contribution to the output variance. It calculates sensitivity indices, such as first-order and total-order indices, which help identify the most influential variables and their interactions. The Sobol method is widely used because it handles nonlinear relationships and interactions between variables.
    - Example paper: [Zhang XY, Trame MN, Lesko LJ, Schmidt S. Sobol Sensitivity Analysis: A Tool to Guide the Development and Evaluation of Systems Pharmacology Models. CPT Pharmacometrics Syst Pharmacol. 2015 Feb;4(2):69-79. doi: 10.1002/psp4.6. PMID: 27548289; PMCID: PMC5006244.](https://pmc.ncbi.nlm.nih.gov/articles/PMC5006244/)
      
* __Morris OAT Method__: The Morris One-At-a-Time (OAT) method is a screening technique that involves varying one input at a time while keeping others constant to estimate the sensitivity of the output to each input. It is useful for screening variables initially to identify which ones have the most significant impact on the model output. The method calculates elementary effects, which provide insights into the importance and nonlinearity of each variable's effect.
    - Example paper: [Adhikari A, Vilkhovoy M, Vadhin S, Lim HE, Varner JD. Effective Biophysical Modeling of Cell-Free Transcription and Translation Processes. Front Bioeng Biotechnol. 2020 Nov 26;8:539081. doi: 10.3389/fbioe.2020.539081. PMID: 33324619; PMCID: PMC7726328.](https://pmc.ncbi.nlm.nih.gov/articles/PMC7726328/)
      
* __eFAST Method__: The Extended Fourier Amplitude Sensitivity Test (eFAST) is another variance-based method that uses the Fourier series to represent the model output. It is computationally efficient and can handle high-dimensional problems by transforming a multivariate problem into a univariate one. eFAST calculates first-order and total-order sensitivity indices, making it suitable for models with many input variables.
    - Example paper: [Xu C, Gertner G. Understanding and comparisons of different sampling approaches for the Fourier Amplitudes Sensitivity Test (FAST). Comput Stat Data Anal. 2011 Jan 1;55(1):184-198. doi: 10.1016/j.csda.2010.06.028. PMID: 24143037; PMCID: PMC3798038.](https://pmc.ncbi.nlm.nih.gov/articles/PMC3798038/)

* __Regression-Based Sensitivity Analysis__: This method uses linear regression to assess the sensitivity of model outputs to input variables. While it is straightforward and can explore the entire space of variables, it is limited by its inability to capture nonlinear dependencies between inputs and outputs effectively.

For a deeper background on _variance sensitivity analysis_, you can refer to the following paper:
* [Atherton, R.W., Schainker, R.B. and Ducot, E.R. (1975), On the statistical sensitivity analysis of models for chemical kinetics. AIChE J., 21: 441-448. https://doi.org/10.1002/aic.690210304](https://aiche.onlinelibrary.wiley.com/doi/abs/10.1002/aic.690210304)
