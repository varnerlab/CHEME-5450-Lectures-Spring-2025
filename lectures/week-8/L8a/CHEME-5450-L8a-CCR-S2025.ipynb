# L8a: The biology of choice. How does a cell choose Glucose?

In this lecture, we'll begin our discussion of biological information processing and choice behavior by exploring a classic example: why/how a cell chooses the six-carbon sugar [Glucose](https://en.wikipedia.org/wiki/Glucose) over other carbon sources. The key concepts of this lecture are:
* [Carbon catabolite repression (CCR)](https://en.wikipedia.org/wiki/Catabolite_repression) is a regulatory mechanism in microorganisms that prioritizes the use of preferred carbon sources (e.g., glucose) by suppressing genes and enzymes needed to metabolize alternative substrates, ensuring efficient resource utilization.
* [Post-translational modifications](https://en.wikipedia.org/wiki/Post-translational_modification) like phosphorylation regulates protein function by inducing structural changes that activate or deactivate enzymatic activity, serving as critical switches in cellular signaling and metabolic pathways. This reversible mechanism allows precise control over processes from cell cycle progression to stress responses while maintaining efficiency through kinase-phosphatase balance.
* [Kinases and phosphatases](https://pubmed.ncbi.nlm.nih.gov/29089159/) are enzymes regulating cellular processes by adding and removing phosphate groups. Kinases transfer phosphate from ATP to substrates (phosphorylation), while phosphatases hydrolyze phosphate from phosphorylated proteins (dephosphorylation), maintaining dynamic regulation of signaling pathways and cellular functions.

## Diauxie and the choice of Glucose 
One of the most famous examples of the choice of carbon sources is the phenomenon of [diauxie](https://en.wikipedia.org/wiki/Diauxie). Diauxie is a biphasic growth pattern observed in bacteria when two carbon sources are available, but one is preferred over the other. 
* The term `diauxie` was coined by Jacques Monod in 1941 to describe the sequential use of two sugars by _Escherichia coli_. This led to the discovery of the [lac operon](https://en.wikipedia.org/wiki/Lac_operon), a set of genes involved in the metabolism of lactose, and later to the [Nobel Prize in Physiology or Medicine in 1965](https://www.nobelprize.org/prizes/medicine/1965/summary/).
